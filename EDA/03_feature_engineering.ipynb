# 03: Feature Engineering Efficacy Proving

In our Kaggle winning solution (v3), we created several composite features. Here, we will visually prove *why* those features work by showing how they separate the target classes.

```
import pandas as pd, numpy as np, seaborn as sns, matplotlib.pyplot as plt
train = pd.read_csv('../../data/train.csv')

# Recreate our v3 Master Features
risk = np.zeros(len(train), dtype='int8')
if 'sleep_duration' in train: risk += (train['sleep_duration'] < 6.0).astype('int8')
if 'stress_level' in train: risk += (train['stress_level'].astype(str) == 'high').astype('int8')
if 'sleep_quality' in train: risk += (train['sleep_quality'].astype(str) == 'poor').astype('int8')
if 'physical_activity_level' in train: risk += (train['physical_activity_level'].astype(str) == 'sedentary').astype('int8')
if 'bmi' in train: risk += (train['bmi'] >= 25.0).astype('int8')
train['lifestyle_risk_index'] = risk

if 'sleep_duration' in train:
    train['sleep_distance_from_8'] = (train['sleep_duration'] - 8.0).abs()
```

## 1. The Power of the Lifestyle Risk Index
Does this composite index actually separate healthy from at-risk students?

```
plt.figure(figsize=(10, 6))
sns.countplot(data=train, x='lifestyle_risk_index', hue='health_condition', palette='magma')
plt.title('Class Separation via Lifestyle Risk Index', fontsize=16, fontweight='bold')
plt.xlabel('Risk Index Score (Higher is Worse)')
plt.ylabel('Number of Students')
plt.show()
```

## 2. Sleep Distance From Ideal (8 hours)
Notice how the further a student deviates from 8 hours of sleep, the higher the risk.

```
plt.figure(figsize=(10, 6))
sns.kdeplot(data=train, x='sleep_distance_from_8', hue='health_condition', fill=True, common_norm=False, palette='crest')
plt.title('KDE Plot: Absolute Distance from 8 Hours of Sleep', fontsize=16)
plt.show()
```
