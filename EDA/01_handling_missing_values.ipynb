# 01: Deep Missing Value Analysis & MNAR Profiling

In this notebook, we move beyond basic counts to understand the **topology of missingness** using the `missingno` library and statistical tests. Our goal is to determine if missingness is MCAR (Missing Completely At Random) or MNAR (Missing Not At Random).

```
!pip install missingno
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import missingno as msno

sns.set_theme(style='whitegrid', context='notebook')
train = pd.read_csv('../../data/train.csv')
print(f'Dataset Shape: {train.shape}')
```

## 1. Missingness Matrix
The missingness matrix allows us to see the distribution of missing values across the entire dataset. Patterns here often indicate MNAR.

```
plt.figure(figsize=(10,6))
msno.matrix(train, sparkline=False, fontsize=12)
plt.title('Missingness Matrix across Features', fontsize=18, y=1.05)
plt.show()
```

## 2. Missingness Correlation Heatmap
Do features tend to be missing together? A high correlation means if one is missing, the other is likely missing too.

```
plt.figure(figsize=(8,6))
msno.heatmap(train, cmap='RdBu', fontsize=12)
plt.title('Nullity Correlation Heatmap', fontsize=18)
plt.show()
```

## 3. Creating Missingness Indicators
Since we suspect missingness carries predictive signal (NMAR), we will create boolean indicators for all columns with nulls, rather than just blindly imputing them.

```
null_cols = train.columns[train.isnull().any()].tolist()
print(f'Columns with missing values: {null_cols}')

for col in null_cols:
    train[f'{col}__is_missing'] = train[col].isnull().astype(int)

print('Successfully created missingness indicators. These will be fed to our GBDT models.')
train[[c for c in train.columns if 'missing' in c]].head()
```
