# 05: Categorical Encoding & Distribution Scaling

Demonstrating the mathematical transformations of our data before feeding into GBDTs.

```
import pandas as pd, numpy as np, seaborn as sns, matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
train = pd.read_csv('../../data/train.csv')
```

## 1. Frequency Encoding (Leak-Free Approximation)
This mirrors our v3 pipeline where we use counts instead of arbitrary ordinal labels.

```
if 'stress_level' in train.columns:
    freq = train['stress_level'].value_counts(dropna=False).to_dict()
    train['stress_level_freq'] = train['stress_level'].map(freq)
    
    plt.figure(figsize=(10, 5))
    sns.histplot(data=train, x='stress_level_freq', hue='health_condition', multiple='stack', palette='tab10')
    plt.title('Frequency Encoded Stress Level vs Target', fontsize=14)
    plt.show()
```

## 2. Min-Max Scaling (Before vs After)

```
if 'bmi' in train.columns:
    scaler = MinMaxScaler()
    train['bmi_scaled'] = scaler.fit_transform(train[['bmi']])
    
    fig, ax = plt.subplots(1, 2, figsize=(14, 5))
    sns.kdeplot(train['bmi'], ax=ax[0], fill=True, color='crimson')
    ax[0].set_title('Original BMI Distribution')
    
    sns.kdeplot(train['bmi_scaled'], ax=ax[1], fill=True, color='teal')
    ax[1].set_title('Min-Max Scaled BMI [0, 1]')
    plt.show()
```
