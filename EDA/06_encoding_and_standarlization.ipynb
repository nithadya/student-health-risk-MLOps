# 06: Standardization (Z-Score) & Robust Scaling

When features have heavy outliers, Z-score (StandardScaler) might skew. We visualize this effect.

```
import pandas as pd, numpy as np, seaborn as sns, matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, RobustScaler
train = pd.read_csv('../../data/train.csv')
```

## 1. StandardScaler vs RobustScaler

```
if 'heart_rate' in train.columns:
    ss = StandardScaler()
    rs = RobustScaler()
    
    train['hr_standard'] = ss.fit_transform(train[['heart_rate']])
    train['hr_robust'] = rs.fit_transform(train[['heart_rate']])
    
    fig, ax = plt.subplots(1, 3, figsize=(18, 5))
    sns.boxplot(y=train['heart_rate'], ax=ax[0], color='lightgreen')
    ax[0].set_title('Original Heart Rate')
    
    sns.boxplot(y=train['hr_standard'], ax=ax[1], color='lightblue')
    ax[1].set_title('StandardScaler (Z-Score)')
    
    sns.boxplot(y=train['hr_robust'], ax=ax[2], color='salmon')
    ax[2].set_title('RobustScaler (IQR-based)')
    plt.tight_layout()
    plt.show()
```
