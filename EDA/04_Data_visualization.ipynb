# 04: Comprehensive Data Visualization & Correlation Topologies

Visualizing high-dimensional relationships.

```
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
import numpy as np
train = pd.read_csv('../../data/train.csv')
```

## 1. Feature Correlation Matrix (Pearson & Spearman)
Understanding monotonic and linear relationships.

```
num_train = train.select_dtypes(include=np.number)
if 'id' in num_train.columns: num_train = num_train.drop(columns=['id'])

fig, axes = plt.subplots(1, 2, figsize=(20, 8))
sns.heatmap(num_train.corr(method='pearson'), annot=True, fmt='.2f', cmap='coolwarm', ax=axes[0], cbar_kws={'shrink': .8})
axes[0].set_title('Pearson Correlation (Linear)', fontsize=16)

sns.heatmap(num_train.corr(method='spearman'), annot=True, fmt='.2f', cmap='viridis', ax=axes[1], cbar_kws={'shrink': .8})
axes[1].set_title('Spearman Correlation (Monotonic)', fontsize=16)
plt.tight_layout()
plt.show()
```

## 2. Multi-Variate Pairplots
How do physiological factors interplay with health conditions?

```
subset = ['age', 'bmi', 'heart_rate', 'sleep_duration', 'health_condition']
actual_cols = [c for c in subset if c in train.columns]

sns.pairplot(train[actual_cols], hue='health_condition', palette='Set1', diag_kind='kde', corner=True)
plt.suptitle('Pairplot of Key Physiological Indicators', y=1.02, fontsize=18)
plt.show()
```

## 3. Categorical Interactions (Stacked Bars)

```
cross_tab = pd.crosstab(train['stress_level'], train['health_condition'], normalize='index')
cross_tab.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='Set2')
plt.title('Proportion of Health Conditions by Stress Level', fontsize=16)
plt.ylabel('Proportion')
plt.show()
```
