# 🌌 Dimensionality Reduction: t-SNE

t-SNE excels at preserving local neighborhoods, making it ideal for finding distinct clusters of students (e.g., finding the "sedentary/high-stress" manifold).

```
import pandas as pd, numpy as np, seaborn as sns, matplotlib.pyplot as plt
import os
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

train = pd.read_csv('../../../data/train.csv')
os.makedirs('../figures', exist_ok=True)
os.makedirs('../data', exist_ok=True)
```

## 1. Compute t-SNE

```
num_cols = train.select_dtypes(include=np.number).columns.tolist()
if 'id' in num_cols: num_cols.remove('id')

X = train[num_cols].fillna(train[num_cols].mean())
X_scaled = StandardScaler().fit_transform(X)

# Subsample for t-SNE speed if dataset is huge, but here we run on all
print('Running t-SNE... (this may take a minute)')
tsne = TSNE(n_components=2, perplexity=30, random_state=42, n_jobs=-1)
tsne_components = tsne.fit_transform(X_scaled)

tsne_df = pd.DataFrame(tsne_components, columns=['TSNE1', 'TSNE2'])
tsne_df['health_condition'] = train['health_condition']

# Save evidence
tsne_df.to_csv('../data/TSNE_components.csv', index=False)
print('✅ t-SNE components saved to ../data/TSNE_components.csv')
```

## 2. t-SNE Visualization Evidence

```
plt.figure(figsize=(10,7))
sns.scatterplot(data=tsne_df, x='TSNE1', y='TSNE2', hue='health_condition', palette='viridis', alpha=0.6)
plt.title('t-SNE Manifold Projection', fontsize=16)

fig_path = '../figures/tsne_health_clusters.png'
plt.savefig(fig_path, dpi=300, bbox_inches='tight')
plt.show()
print(f'✅ Figure saved to {fig_path}')
```
