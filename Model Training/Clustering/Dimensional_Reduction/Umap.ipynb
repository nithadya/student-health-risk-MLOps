# 🕳️ Dimensionality Reduction: UMAP

UMAP is faster than t-SNE and preserves global structure better.

```
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
import os, sys

# Check if umap is installed
try:
    import umap
except ImportError:
    import subprocess
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', 'umap-learn'])
    import umap

from sklearn.preprocessing import StandardScaler
train = pd.read_csv('../../../data/train.csv')
os.makedirs('../figures', exist_ok=True)
os.makedirs('../data', exist_ok=True)
```

## 1. Compute UMAP

## 2. UMAP Visualization Evidence

```
plt.figure(figsize=(10,7))
sns.scatterplot(data=umap_df, x='UMAP1', y='UMAP2', hue='health_condition', palette='coolwarm', alpha=0.6)
plt.title('UMAP Projection', fontsize=16)

fig_path = '../figures/umap_projection.png'
plt.savefig(fig_path, dpi=300, bbox_inches='tight')
plt.show()
print(f'✅ Figure saved to {fig_path}')
```
