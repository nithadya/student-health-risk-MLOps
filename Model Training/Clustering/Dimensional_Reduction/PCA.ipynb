# 🧊 Dimensionality Reduction: PCA

Principal Component Analysis (PCA) maps our high-dimensional health data into 2D/3D space, maximizing variance. This helps us visualize if the classes (`health_condition`) naturally separate.

```
import pandas as pd, numpy as np, seaborn as sns, matplotlib.pyplot as plt
import os
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

train = pd.read_csv('../../../data/train.csv')
os.makedirs('../figures', exist_ok=True)
os.makedirs('../data', exist_ok=True)
```

## 1. Compute PCA

```
num_cols = train.select_dtypes(include=np.number).columns.tolist()
if 'id' in num_cols: num_cols.remove('id')

# Impute and scale
X = train[num_cols].fillna(train[num_cols].mean())
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2, random_state=42)
pca_components = pca.fit_transform(X_scaled)

print(f'Explained Variance Ratio: {pca.explained_variance_ratio_}')

pca_df = pd.DataFrame(pca_components, columns=['PC1', 'PC2'])
pca_df['health_condition'] = train['health_condition']

# Save evidence
pca_df.to_csv('../data/PCA__components.csv', index=False)
print('✅ PCA components saved to ../data/PCA__components.csv')
```

## 2. PCA Visualization Evidence

```
plt.figure(figsize=(10,7))
sns.scatterplot(data=pca_df, x='PC1', y='PC2', hue='health_condition', palette='magma', alpha=0.6)
plt.title('2D PCA Projection of Health Data', fontsize=16)

fig_path = '../figures/pca_2d_scatter.png'
plt.savefig(fig_path, dpi=300, bbox_inches='tight')
plt.show()
print(f'✅ Figure saved to {fig_path}')
```
