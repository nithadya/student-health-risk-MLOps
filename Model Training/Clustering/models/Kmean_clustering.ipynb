# 🎯 K-Means Clustering

We apply unsupervised K-Means to our PCA components to discover if the mathematical clusters align with the actual `health_condition` categories.

```
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt, os
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

os.makedirs('../figures', exist_ok=True)

if not os.path.exists('../data/PCA__components.csv'):
    raise FileNotFoundError('Run PCA.ipynb first to generate components!')
    
pca_df = pd.read_csv('../data/PCA__components.csv')
X_pca = pca_df[['PC1', 'PC2']]
```

## 1. Train K-Means (K=3)

```
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_pca)

pca_df['KMeans_Cluster'] = clusters.astype(str)

sil_score = silhouette_score(X_pca, clusters)
print(f'Silhouette Score: {sil_score:.4f}')
```

## 2. K-Means Visualization Evidence

```
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot K-Means Clusters
sns.scatterplot(data=pca_df, x='PC1', y='PC2', hue='KMeans_Cluster', palette='Set1', ax=axes[0], alpha=0.6)
axes[0].scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=200, c='black', marker='X', label='Centroids')
axes[0].set_title('K-Means Discovered Clusters (K=3)', fontsize=14)
axes[0].legend()

# Plot Actual Conditions
sns.scatterplot(data=pca_df, x='PC1', y='PC2', hue='health_condition', palette='magma', ax=axes[1], alpha=0.6)
axes[1].set_title('Actual Health Conditions', fontsize=14)

fig_path = '../figures/kmeans_clusters_comparison.png'
plt.savefig(fig_path, dpi=300, bbox_inches='tight')
plt.show()
print(f'✅ Figure saved to {fig_path}')
```
