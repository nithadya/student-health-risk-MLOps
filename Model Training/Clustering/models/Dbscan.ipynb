# 📡 DBSCAN Density Clustering

DBSCAN is excellent for finding arbitrary shaped clusters and detecting outliers (noise) in our t-SNE manifold.

```
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt, os
from sklearn.cluster import DBSCAN

os.makedirs('../figures', exist_ok=True)

if not os.path.exists('../data/TSNE_components.csv'):
    raise FileNotFoundError('Run T-sne.ipynb first to generate components!')
    
tsne_df = pd.read_csv('../data/TSNE_components.csv')
X_tsne = tsne_df[['TSNE1', 'TSNE2']]
```

## 1. Train DBSCAN

```
dbscan = DBSCAN(eps=2.5, min_samples=10)
clusters = dbscan.fit_predict(X_tsne)

tsne_df['DBSCAN_Cluster'] = clusters.astype(str)

num_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)
num_noise = list(clusters).count(-1)
print(f'Estimated number of clusters: {num_clusters}')
print(f'Estimated number of noise points (outliers): {num_noise}')
```

## 2. DBSCAN Visualization Evidence

```
plt.figure(figsize=(10, 7))
sns.scatterplot(data=tsne_df, x='TSNE1', y='TSNE2', hue='DBSCAN_Cluster', palette='tab20', alpha=0.7)
plt.title('DBSCAN Density Clusters on t-SNE Manifold (Black = Outliers)', fontsize=16)

fig_path = '../figures/dbscan_outliers.png'
plt.savefig(fig_path, dpi=300, bbox_inches='tight')
plt.show()
print(f'✅ Figure saved to {fig_path}')
```
